```python
from torchvision import transforms
from torchvision.datasets import ImageFolder

from PIL import Image

import os

# =====================================================
# AUGMENTATION TRANSFORM
# =====================================================

train_transform = transforms.Compose([

    transforms.Resize((256, 256)),

    transforms.RandomResizedCrop(224),

    transforms.RandomHorizontalFlip(p=0.5),

    transforms.RandomRotation(20),

    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    )

])

# =====================================================
# LOAD TRAIN DATASET
# =====================================================

train_dataset = ImageFolder(
    root=r'C:\Desktop\Dog_Cat_Cls\dataset\train'
)

# =====================================================
# OUTPUT FOLDER
# =====================================================

output_dir = "augmented_dataset"

os.makedirs(output_dir, exist_ok=True)

# =====================================================
# CREATE CLASS FOLDERS
# =====================================================

for class_name in train_dataset.classes:

    class_folder = os.path.join(
        output_dir,
        class_name
    )

    os.makedirs(class_folder, exist_ok=True)

# =====================================================
# NUMBER OF AUGMENTATIONS PER IMAGE
# =====================================================

augmentations_per_image = 2

# =====================================================
# GENERATE & SAVE AUGMENTED IMAGES
# =====================================================

for image_path, label in train_dataset.samples:

    # Get class name
    class_name = train_dataset.classes[label]

    # Open image
    image = Image.open(image_path).convert("RGB")

    # Original filename
    filename = os.path.splitext(
        os.path.basename(image_path)
    )[0]

    # =================================================
    # GENERATE MULTIPLE AUGMENTATIONS
    # =================================================

    for i in range(augmentations_per_image):

        # Apply augmentation
        augmented_image = train_transform(image)

        # Save path
        save_path = os.path.join(
            output_dir,
            class_name,
            f"{filename}_aug_{i}.jpg"
        )

        # Save image
        augmented_image.save(save_path)

print("Augmented dataset created successfully!")
```
